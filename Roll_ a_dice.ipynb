```python
import random

def dice_rolling():
    a=random.randint(1, 6)
    b=random.randint(1, 6)

    return(a, b)


dice_input = input("Roll a dice? (y/n): ").lower()
if dice_input == "y":
    print(dice_rolling())
elif dice_input == "n":
    print("Thanks for playing")
else:
    print("Invalid choice!")

```
